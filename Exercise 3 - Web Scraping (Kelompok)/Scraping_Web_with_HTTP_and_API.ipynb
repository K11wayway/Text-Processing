**Scraping banyak pages**

```python
import cloudscraper
from bs4 import BeautifulSoup
import time

root = 'https://www.historia.id/articles/categories/politik/page/'
scraper = cloudscraper.create_scraper(browser={'browser': 'chrome', 'platform': 'windows', 'mobile': False})

kategori_judul = []
links = []
penjelasan = []

for page in range(1, 6):
    print(f"Mengambil data dari Halaman {page}...")

    url = root + str(page)
    response = scraper.get(url)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')

        semua_judul_mentah = soup.find_all(['h2', 'h3'])
        judul_valid = [tag for tag in semua_judul_mentah
                       if tag.text.strip() not in ['Ikuti Kami', 'TERPOPULER', 'TRENDING', 'TRENDING :']
                       and not tag.text.strip().isdigit()
                       and tag.text.strip() != ""]

        semua_desc = soup.find_all('p', class_=lambda x: x and 'wixui-collapsible-text' in x)

        for i, tag_judul in enumerate(judul_valid):
            teks_judul = tag_judul.text.strip()

            kategori_judul.append(teks_judul)

            link_tag = tag_judul.find_parent('a') or tag_judul.find('a')
            if link_tag and 'href' in link_tag.attrs:
                href = link_tag['href']
                full_url = href if href.startswith('http') else 'https://www.historia.id' + href
                links.append(full_url)
            else:
                links.append("-")

            if i < len(semua_desc):
                penjelasan.append(semua_desc[i].text.strip())
            else:
                penjelasan.append("-")

    else:
        print(f"Gagal mengakses halaman {page}. Status Code: {response.status_code}")

    time.sleep(2)

print("Scraping donee")
print(f"   Halaman {page}: {len(judul_valid)} judul, {len(semua_desc)} deskripsi ditemukan")
```

```python
import pandas as pd
df = pd.DataFrame({'Judul Artikel': kategori_judul,
                   'Link Halaman': links,
                   'Penjelasan': penjelasan})
df
```

```python
# Berikut untuk untuk deskripsi yang konsisten dan lebih informatif dengan menggunakan <meta name="description">

import time

deskripsi_lengkap = []

for i, link in enumerate(links):
    if link == "-":
        deskripsi_lengkap.append("-")
        continue

    try:
        resp = scraper.get(link)
        if resp.status_code == 200:
            soup_artikel = BeautifulSoup(resp.text, 'html.parser')

            meta_desc = soup_artikel.find('meta', attrs={'name': 'description'})
            if not meta_desc or not meta_desc.get('content'):
                meta_desc = soup_artikel.find('meta', attrs={'property': 'og:description'})

            if meta_desc and meta_desc.get('content'):
                deskripsi_lengkap.append(meta_desc['content'].strip())
            else:
                deskripsi_lengkap.append("-")
        else:
            deskripsi_lengkap.append("-")
    except Exception as e:
        deskripsi_lengkap.append("-")

    print(f"{i+1}/{len(links)} selesai: {kategori_judul[i][:50]}")
    time.sleep(1.5)

df['Penjelasan'] = deskripsi_lengkap
df
```

```python
import requests

url = "https://api.coingecko.com/api/v3/coins/markets"
params = {
    "vs_currency": "idr",
    "per_page": 5
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()
    for coin in data:
        print(f"Nama Kripto : {coin['name']}")
        print(f"Harga IDR   : Rp {coin['current_price']:,}")
        print("-" * 30)
else:
    print("Gagal ngambil data")
```
